## Instructions

Your task is to write a program that:

1. Asks the user for their **name** and **address**
2. Asks for the room dimensions (in this order):
   - Square room: one side length
   - Rectangular room: length and width
   - Circular room: diameter
3. Calculates the area of each room
4. Calculates the material and installation cost using:
   - `MATERIAL_COST = 3.25`
   - `INSTALLATION_COST = 2.10`
5. Displays a nicely formatted estimate for each room
6. Shows the **grand total**

### Requirements

- Use `float()` instead of `eval()` to handle decimal input

````python
# Author: Nathacha Marc
# Description:
#   Flooring estimate calculator for three rooms:
#   - Square room
#   - Rectangular room
#   - Circular room
#   The program collects customer information and room dimensions,
#   computes areas, material and installation costs, and prints
#   a nicely formatted estimate including a room summary table
#   and grand total.

import math  # math module used for circular room area (pi)

# ----------------------------------------------------------
# CONSTANTS
# ----------------------------------------------------------
MATERIAL_COST = 3.25       # Cost per square foot for material
INSTALLATION_COST = 2.10   # Cost per square foot for installation
TOTAL_RATE = MATERIAL_COST + INSTALLATION_COST  # Combined rate

# ----------------------------------------------------------
# CUSTOMER INFORMATION
# ----------------------------------------------------------
name = input("Enter your name: ")
address = input("Enter your address: ")

print("\n--- Room Dimension Entry ---")

# ----------------------------------------------------------
# ROOM DIMENSIONS (all inputs use float() for decimal support)
# ----------------------------------------------------------
# Square room: one side
square_side = float(input("Enter the side length of the SQUARE room (in feet): "))

# Rectangular room: length and width
rect_length = float(input("Enter the length of the RECTANGULAR room (in feet): "))
rect_width = float(input("Enter the width of the RECTANGULAR room (in feet): "))

# Circular room: diameter
circle_diameter = float(input("Enter the diameter of the CIRCULAR room (in feet): "))

# ----------------------------------------------------------
# AREA CALCULATIONS
# ----------------------------------------------------------
square_area = square_side ** 2
rect_area = rect_length * rect_width
circle_area = math.pi * (circle_diameter / 2) ** 2  # uses math.pi as required

# ----------------------------------------------------------
# COST CALCULATIONS
# ----------------------------------------------------------
# Square room
square_material_cost = square_area * MATERIAL_COST
square_install_cost = square_area * INSTALLATION_COST
square_total_cost = square_area * TOTAL_RATE

# Rectangular room
rect_material_cost = rect_area * MATERIAL_COST
rect_install_cost = rect_area * INSTALLATION_COST
rect_total_cost = rect_area * TOTAL_RATE

# Circular room
circle_material_cost = circle_area * MATERIAL_COST
circle_install_cost = circle_area * INSTALLATION_COST
circle_total_cost = circle_area * TOTAL_RATE

# Grand total for all rooms
grand_total = square_total_cost + rect_total_cost + circle_total_cost

# ----------------------------------------------------------
# OUTPUT: DETAILED ESTIMATE
# ----------------------------------------------------------
print("\n================ FLOORING ESTIMATE ================")
print(f"Customer: {name}")
print(f"Address:  {address}")
print("==================================================")

# Optional but encouraged: table-style summary for all rooms
print("\nRoom Summary")
print("--------------------------------------------------")
print(f"{'Room Type':<15}{'Area (sq ft)':>12}{'Material ($)':>15}{'Install ($)':>15}{'Total ($)':>12}")
print("-" * 69)
print(f"{'Square':<15}{square_area:>12.2f}{square_material_cost:>15.2f}{square_install_cost:>15.2f}{square_total_cost:>12.2f}")
print(f"{'Rectangular':<15}{rect_area:>12.2f}{rect_material_cost:>15.2f}{rect_install_cost:>15.2f}{rect_total_cost:>12.2f}")
print(f"{'Circular':<15}{circle_area:>12.2f}{circle_material_cost:>15.2f}{circle_install_cost:>15.2f}{circle_total_cost:>12.2f}")
print("--------------------------------------------------")

# More detailed breakdown per room (matches typical sample layouts)
print("\nSQUARE ROOM")
print(f"  Area:                  {square_area:.2f} sq ft")
print(f"  Material Cost:        ${square_material_cost:.2f}")
print(f"  Installation Cost:    ${square_install_cost:.2f}")
print(f"  Total Room Cost:      ${square_total_cost:.2f}")
print("--------------------------------------------------")

print("RECTANGULAR ROOM")
print(f"  Area:                  {rect_area:.2f} sq ft")
print(f"  Material Cost:        ${rect_material_cost:.2f}")
print(f"  Installation Cost:    ${rect_install_cost:.2f}")
print(f"  Total Room Cost:      ${rect_total_cost:.2f}")
print("--------------------------------------------------")

print("CIRCULAR ROOM")
print(f"  Area:                  {circle_area:.2f} sq ft")
print(f"  Material Cost:        ${circle_material_cost:.2f}")
print(f"  Installation Cost:    ${circle_install_cost:.2f}")
print(f"  Total Room Cost:      ${circle_total_cost:.2f}")
print("--------------------------------------------------")

# Grand total for all rooms
print(f"GRAND TOTAL:            ${grand_total:.2f}")
print("==================================================")
````
