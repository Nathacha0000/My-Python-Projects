````python
# Name: Nathacha Marc
# Project: Beat The Bot (Rock Paper Scissors)
# Date: 02/09/2026

import random   # needed for the bot's random moves

# -----------------------------------
# Welcome message
# -----------------------------------
print("Welcome to Beat the Bot!")

# -----------------------------------
# Ask user for number of rounds
# Must be odd and between 3 and 11
# -----------------------------------
rounds_valid = False

while rounds_valid == False:
    user_rounds = input("How many rounds would you like to play? (odd number between 3 and 11): ")

    # check if number
    if user_rounds.isdigit() == False:
        print("Invalid input. Please enter a number.")
        continue

    user_rounds = int(user_rounds)

    # check allowed range + odd
    if user_rounds < 3 or user_rounds > 11 or user_rounds % 2 == 0:
        print("That’s not a valid number. Please enter an ODD number between 3 and 11.")
    else:
        rounds_valid = True   # number is good

total_rounds = user_rounds

# -----------------------------------
# lists to store game info
# -----------------------------------
player_moves = []
bot_moves = []
round_outcomes = []

# counters for final summary
player_wins = 0
bot_wins = 0
ties = 0

# move options
move_codes = ["R", "P", "S"]
move_words = ["Rock", "Paper", "Scissors"]

# -----------------------------------
# main game loop
# -----------------------------------
current_round = 1

while current_round <= total_rounds:
    print()
    print("Round", current_round)

    # ask player for move
    move_ok = False
    while move_ok == False:
        player_input = input("Enter your move (R)ock, (P)aper, or (S)cissors: ")
        player_input = player_input.strip().upper()

        if player_input not in move_codes:
            print("Invalid choice. Please enter R, P, or S.")
        else:
            move_ok = True

    # convert to full name
    player_index = move_codes.index(player_input)
    player_word = move_words[player_index]

    # bot move
    bot_code = random.choice(move_codes)
    bot_index = move_codes.index(bot_code)
    bot_word = move_words[bot_index]

    print("You chose:", player_word)
    print("Bot chose:", bot_word)

    # -----------------------------------
    # determine winner
    # -----------------------------------
    if player_input == bot_code:
        outcome = "Tie"
        print("It's a TIE.")
        ties = ties + 1
    elif (player_input == "R" and bot_code == "S") or \
         (player_input == "P" and bot_code == "R") or \
         (player_input == "S" and bot_code == "P"):
        outcome = "Win"
        print("You WIN this round!")
        player_wins = player_wins + 1
    else:
        outcome = "Lose"
        print("You LOSE this round.")
        bot_wins = bot_wins + 1

    # store info in lists
    player_moves.append(player_word)
    bot_moves.append(bot_word)
    round_outcomes.append(outcome)

    current_round = current_round + 1

# -----------------------------------
# summary section
# -----------------------------------
print()
print("Game over! Here's your battle summary:")
print()
print("Round   You        Bot        Result")
print("----------------------------------------")

# print one line per round
for i in range(total_rounds):
    print(f"{i+1:<7}{player_moves[i]:<10}{bot_moves[i]:<10}{round_outcomes[i]}")

# final stats
print()
print("Summary:")
print("Your wins:", player_wins)
print("Bot wins:", bot_wins)
print("Ties:", ties)

print()
print("Thanks for playing!")
````
