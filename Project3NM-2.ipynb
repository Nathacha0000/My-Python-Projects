# Sample Starter Program

In the code cell below, a starter project has been provided for you.

This includes:
- a main() function that contains the main program loop
- a sample data dictionary
- several implemented helper functions
- placeholders for the functions you are expected to implement

 The purpose of this structure is to help you focus on applying core programming skills—such as working with dictionaries, writing functions, and validating input—without getting overwhelmed by program design. Do not modify the program structure or the location of the main dictionary. Doing so may affect grading and make it harder to test your code.

````python

# Name: Nathacha Marc
# MSIT 501 - Project 3
# Date: 02/23/2026
# Movie Lover's Club
# You MUST use the provided dictionary to receive full credit
# You MUST use the starter code below to receive full credit

# You can use the following code as a starting point:

def display_menu():
    """
    Displays the main menu options for the Movie Lover's Club application.

    Parameters: None
    Returns: None

    This function simply prints the available user actions. It does not handle any logic or input.

    You should not have to modify this code.
    """
    print("\nMain Menu")
    print("1. Display all members")
    print("2. Display all movie information for a member")
    print("3. Increment the times a specific movie was watched by a member")
    print("4. Add a movie for a member")
    print("5. Add a new member")
    print("6. Search for a movie across all members")
    print("Q. Quit")


# I've already implemented three of the input helper functions to help you get started

def get_existing_member_name(movies):
    """
    Prompts the user to enter a member name and checks if it exists in the movies dictionary.

    Parameters:
        movies (dict): The main dictionary containing members and their movie data.

    Returns:
        str or None: The validated member name if it exists, or None if not found.

    You should not have to modify this code.
    """
    name = input("Enter the member's name: ").strip().title()

    # Check if member name is in the dictionary
    if name in movies:
        return name
    else:
        print("Sorry, member not found.")
        return None


def get_existing_movie_name(movie_dict):
    """
    Prompts the user to enter a movie title and checks if it exists in the member's movie dictionary.

    Parameters:
        movie_dict (dict): The dictionary containing a single member's movies.

    Returns:
        str or None: The validated movie title if it exists, or None if not found.

    You should not have to modify this code.
    """
    title = input("Enter the movie title: ").strip().title()

    # Check if the movie exists in this member's movie list
    if title in movie_dict:
        return title
    else:
        print("Sorry, movie title not found.")
        return None


def get_movie_search_title():
    """
    Prompts the user to enter the title of a movie to search for across all members.

    Parameters:
        None

    Returns:
        str: The entered movie to be used in search

    """
    title = input("Enter the movie title to search for: ").strip()
    return title


# Placeholder function definitions to allow the code to run
# You must implement these functions

def display_all_members(movies):
    # Display all member names from the dictionary
    print("\nClub Members")
    print("===================")
    for member_name in movies:
        print(member_name)


def display_movies_for_member(movies, name):
    # Display all movies for the specified member, with rating and watch count
    # Format this information nicely in a table
    member_movies = movies.get(name, {})

    # If the member has no movies yet
    if not member_movies:
        print(f"\n{name} has not added any movies yet.")
        return

    print(f"\nMovies for club member: {name}")
    print("Movie                    Rating   Watched")
    print("=============================================")

    # Loop through movies and print info
    for title, info in member_movies.items():
        rating = info.get("Rating", "")
        watched = info.get("Watched", 0)
        # Title left-aligned, rating left-aligned, watched right-aligned
        print(f"{title:<24}{rating:<9}{watched:>6}")


def increment_watch_count(movies, name, title):
    # Increase the watch count by 1 for a specific movie for a specific member
    movies[name][title]["Watched"] += 1
    print("Times watched incremented.")


def get_new_movie_info(movie_dict):
    # This function prompts the user to enter information for a new movie, including:
    # - the movie title (must not already exist in the member's movie list)
    # - the number of times watched (must be a non-negative integer)
    # - the movie rating (must not be empty)
    # It returns a tuple: (title, times, rating) if all inputs are valid.
    # If any input is invalid or the movie already exists, it returns None.
    # Be efficient here.
    # - If the movie title entered already exists, do not ask for further information
    # - If the times watched entered is illegal, do not ask for the rating

    # Ask for title first
    title = input("Enter the movie title: ").strip().title()

    # Check for duplicate movie
    if title in movie_dict:
        print("That movie already exists for this member.")
        return None

    # Ask for times watched
    times_input = input("Enter how many times it was watched: ").strip()
    if not times_input.isdigit():
        print("Invalid input for times watched. Please enter a non-negative integer.")
        return None

    times = int(times_input)
    if times < 0:
        print("Times watched cannot be negative.")
        return None

    # Ask for rating
    rating = input("Enter the rating (e.g., G, PG, PG-13, R): ").strip()
    if rating == "":
        print("Rating cannot be empty.")
        return None

    return (title, times, rating)


def add_movie_for_member(movies, name, title, times, rating):
    # Add a new movie to the member's movies with the given info
    if name not in movies:
        # Extra safety check
        print("Member not found.")
        return

    movies[name][title] = {
        "Watched": times,
        "Rating": rating
    }
    print("Movie added.")


def get_new_member_name(movies):
    # Prompt user for a new member name and ensure it doesn't already exist
    name = input("Enter the new member's name: ").strip().title()

    if name == "":
        print("Member name cannot be empty.")
        return None

    if name in movies:
        print("That member already exists.")
        return None

    return name


def add_new_member(movies, name):
    # Add the new member to the movies dictionary
    movies[name] = {}
    print("Member added.")


def search_movie_across_members(movies, title):
    # Search for the given movie title across all members and display matches
    search_title = title.strip().title()
    results = []

    for member_name, movie_dict in movies.items():
        if search_title in movie_dict:
            watched = movie_dict[search_title].get("Watched", 0)
            results.append((member_name, watched))

    if not results:
        print(f"\nNo members have watched '{search_title}'.")
        return

    print(f"\nMembers who have watched '{search_title}':\n")
    print("Member                Watched")
    print("------------------------------")
    for member_name, watched in results:
        print(f"{member_name:<22}{watched:>6}")


def main():
    """
    Main function that runs the Movie Lover's Club program.

    This function displays the menu and handles user input.
    It calls specific helper and action functions based on the menu choice.

    You should NOT need to make any changes to the main() function.
    """
    movies = {
        'Mary': {
            'Big': {'Watched': 1, 'Rating': 'G'},
            'Superman': {'Watched': 3, 'Rating': 'PG'},
            'Forrest Gump': {'Watched': 3, 'Rating': 'PG-13'}
        },
        'Frank': {
            'Beauty and the Beast': {'Watched': 1, 'Rating': 'G'},
            'Kung Fu Panda': {'Watched': 5, 'Rating': 'G'},
            'Cinderella': {'Watched': 1, 'Rating': 'G'}
        },
        'Kevin': {
            'Cinderella': {'Watched': 2, 'Rating': 'G'},
            'Superman': {'Watched': 1, 'Rating': 'PG'},
            'Inception': {'Watched': 4, 'Rating': 'PG-13'}
        }
    }

    print("Welcome to the Movie Lover's Club")

    # Main program loop - displays the menu and processes user selections
    # You should not have to modify the main() function
    while True:
        display_menu()
        selection = input("Please enter a selection: ").strip().upper()

        if selection == '1':
            display_all_members(movies)
        elif selection == '2':
            name = get_existing_member_name(movies)
            if name:
                display_movies_for_member(movies, name)
        elif selection == '3':
            name = get_existing_member_name(movies)
            if name:
                title = get_existing_movie_name(movies[name])
                if title:
                    increment_watch_count(movies, name, title)
        elif selection == '4':
            name = get_existing_member_name(movies)
            if name:
                result = get_new_movie_info(movies[name])
                if result:
                    title, times, rating = result
                    add_movie_for_member(movies, name, title, times, rating)
        elif selection == '5':
            name = get_new_member_name(movies)
            if name:
                add_new_member(movies, name)
        elif selection == '6':
            title = get_movie_search_title()
            search_movie_across_members(movies, title)
        elif selection == 'Q':
            print("Thanks for using the Movie Lover's Club application. Goodbye.")
            break
        else:
            print("Invalid selection. Please choose a valid menu option.")

# Call the main function to start your program
if __name__ == "__main__":
    main()
````
